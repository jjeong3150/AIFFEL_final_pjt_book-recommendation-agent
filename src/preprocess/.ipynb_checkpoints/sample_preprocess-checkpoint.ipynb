# 서울 도서관 도서 목록 전처리
`seoul_library_book_list_202604.csv` → `data/processed/seoul_books.csv`

```python
import pandas as pd
import numpy as np

RAW_PATH = "../../data/raw/seoul_library_book_list_202604.csv"
OUT_PATH  = "../../data/processed/seoul_books.csv"
```

## 1. 로드 및 기본 확인

```python
# utf-8-sig: BOM 문자 자동 처리
df = pd.read_csv(RAW_PATH, encoding="utf-8-sig")

# 마지막 열이 빈 컬럼이면 제거
df = df.loc[:, ~df.columns.str.fullmatch(r"Unnamed.*")]

print("shape:", df.shape)
print("columns:", df.columns.tolist())
df.head(3)
```

```python
# 결측값 현황
df.isnull().sum().sort_values(ascending=False)
```

```python
df.dtypes
```

## 2. 컬럼 정리

```python
df = df.rename(columns={
    "번호":        "no",
    "도서명":      "title",
    "저자":        "author",
    "출판사":      "publisher",
    "발행년도":    "publish_year",
    "ISBN":        "isbn",
    "세트 ISBN":   "isbn_set",
    "부가기호":    "addition_symbol",
    "권":          "volume",
    "주제분류번호": "class_no",
    "도서권수":    "book_count",
    "대출건수":    "loan_count",
    "등록일자":    "registered_date",
})

df.head(2)
```

## 3. 타입 변환

```python
df["publish_year"]    = pd.to_numeric(df["publish_year"], errors="coerce").astype("Int64")
df["loan_count"]      = pd.to_numeric(df["loan_count"],   errors="coerce").fillna(0).astype(int)
df["registered_date"] = pd.to_datetime(df["registered_date"], errors="coerce")

df.dtypes
```

## 4. 필터링

```python
print("전처리 전:", len(df))

# ISBN 없는 행 제거
df = df[df["isbn"].notna() & (df["isbn"].astype(str).str.strip() != "")]
print("ISBN 없는 행 제거 후:", len(df))

# 제목 없는 행 제거
df = df[df["title"].notna() & (df["title"].astype(str).str.strip() != "")]
print("제목 없는 행 제거 후:", len(df))
```

## 5. 중복 처리

```python
# ISBN 기준 중복 확인
dup_count = df.duplicated(subset="isbn").sum()
print(f"ISBN 중복 행: {dup_count}건")

# 중복 시 대출건수 합산 후 첫 번째 행 유지
loan_sum = df.groupby("isbn")["loan_count"].sum().rename("loan_count_total")
df = df.drop_duplicates(subset="isbn", keep="first").merge(loan_sum, on="isbn")
df["loan_count"] = df["loan_count_total"]
df = df.drop(columns=["loan_count_total"])

print("중복 제거 후:", len(df))
df.head(3)
```

## 6. 저장

```python
import os
os.makedirs(os.path.dirname(OUT_PATH), exist_ok=True)

df.to_csv(OUT_PATH, index=False, encoding="utf-8-sig")
print(f"저장 완료 -> {OUT_PATH}")
print(f"최종 행 수: {len(df)}")
```
